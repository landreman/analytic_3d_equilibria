Check that $\nabla\cdot B=0$, $B\cdot\nabla\psi=0$, and $(\nabla\times B)\times
B = \nabla p$ for the sheared-iota family directly from the explicit formulas
for $B, \psi$, and $p$ using sympy.

```python
import sympy as sp
```

```python
from sympy import I, pi
x, y, z = sp.symbols('x y z', real=True)
eps, lambd, S, pa = sp.symbols("eps lambd S pa", positive=True)

# Sympy had an easier time simplifying the results if the real and imaginary
# parts were computed this way instead of using its built-in real and imag functions.
def real(s):
    return (s + s.subs(I, -I)) / 2

def imag(s):
    return (s - s.subs(I, -I)) / (2 * I)


# Mutation checks: if any of these values is different from 1,
# the checks should fail.
mutation_1 = 1
mutation_2 = 1
mutation_3 = 1

omega = x + mutation_1 * I * y
omegabar = x - mutation_1 * I * y
K = omegabar * sp.sqrt(1 + eps / omegabar**2)
Xi = omega * K + pi / 2 - S
Bx = real(I * sp.exp(-I * lambd * z) * sp.sin(Xi) / (2 * K)) * mutation_2
By = imag(I * sp.exp(-I * lambd * z) * sp.sin(Xi) / (2 * K))
Bz = (1 / lambd) * real(sp.exp(-I * lambd * z) * sp.cos(Xi)) * mutation_3
psi = (sp.sin(lambd * z)**2 + real(sp.exp(-I * lambd * z) * sp.cos(Xi))**2) / 2
p = pa - (1 / lambd)**2 * psi
```

All derivatives below are taken with respect to the Cartesian coordinates $x$,
$y$, $z$. To help sympy simplify the results, it proved helpful to change variables
from $(x, y)$ to $w = x + iy$ and $\bar w = x - iy$, write $\sin$ and $\cos$ as exponentials, expand, and then simplify.
A result of 0 shows that the original expression vanishes.

```python
w, wb = sp.symbols("w wb")

def simplify(expr):
    expr2 = expr.subs({x: (w + wb) / 2, y: (w - wb) / (2 * I)})
    return sp.simplify(sp.expand(expr2.rewrite(sp.exp)))
```

```python
divB = sp.diff(Bx, x) + sp.diff(By, y) + sp.diff(Bz, z)
simplify(divB)
```

```python
B_dot_grad_psi = Bx * sp.diff(psi, x) + By * sp.diff(psi, y) + Bz * sp.diff(psi, z)
simplify(B_dot_grad_psi)
```

```python
Jx = sp.diff(Bz, y) - sp.diff(By, z)
Jy = sp.diff(Bx, z) - sp.diff(Bz, x)
Jz = sp.diff(By, x) - sp.diff(Bx, y)
dpdx = sp.diff(p, x)
dpdy = sp.diff(p, y)
dpdz = sp.diff(p, z)
```

```python
# x component
residual_x = Jy * Bz - Jz * By - dpdx
simplify(residual_x)
```

```python
# y component
residual_y = Jz * Bx - Jx * Bz - dpdy
simplify(residual_y)
```

```python
# z component
residual_z = Jx * By - Jy * Bx - dpdz
simplify(residual_z)
```
