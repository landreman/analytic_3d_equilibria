```python
import sympy as sp
from sympy.abc import x, y, z
```

```python
eps, pa = sp.symbols("eps pa")
a = sp.sqrt(1 + eps)
b = sp.sqrt(1 - eps)
X = x / a
Y = y / b
s = X**2 + Y**2
F = sp.sqrt(2*s - s*s - 4 * z * z)
Bx = (2 * z * x - (a / b) * F * y) / s
By = (2 * z * y + (b / a) * F * x) / s
Bz = 1 - s
modB2 = Bx*Bx + By*By + Bz*Bz
psi = (x*x + y*y + 4*z*z + modB2 - 2 + eps**2) / 4
p = pa - 2 * psi
```

Check $\nabla\cdot\mathbf{B}=0$

```python
divB = sp.diff(Bx, x) + sp.diff(By, y) + sp.diff(Bz, z)
sp.simplify(divB)
```

Check $\mathbf{B}\cdot\nabla\psi = 0$

```python
B_dot_grad_psi = Bx * sp.diff(psi, x) + By * sp.diff(psi, y) + Bz * sp.diff(psi, z)
sp.simplify(B_dot_grad_psi)
```

Check force balance

```python
Jx = sp.diff(Bz, y) - sp.diff(By, z)
Jy = sp.diff(Bx, z) - sp.diff(Bz, x)
Jz = sp.diff(By, x) - sp.diff(Bx, y)
dpdx = sp.diff(p, x)
dpdy = sp.diff(p, y)
dpdz = sp.diff(p, z)
```

```python
# x component
residual_x = Jy * Bz - Jz * By - dpdx
sp.simplify(residual_x)
```

```python
# y component
residual_y = Jz * Bx - Jx * Bz - dpdy
sp.simplify(residual_y)
```

```python
# z component
residual_z = Jx * By - Jy * Bx - dpdz
sp.simplify(residual_z)
```
